# Análisis y Predicción de la Popularidad de Canciones en Spotify

# 1. Comprensión del Negocio

## 1.1 Descripción del proyecto
El presente proyecto desarrolla un modelo capaz de predecir la popularidad de canciones en Spotify a partir de sus características musicales. Para ello, se emplea un conjunto de datos obtenido de Kaggle y se sigue la metodología CRISP-DM.

De manera complementaria, se realiza un análisis mediante técnicas de aprendizaje no supervisado para identificar grupos de canciones con características similares. Finalmente, se desarrolla una aplicación interactiva utilizando Shiny para Python, la cual permite visualizar los principales resultados del proyecto y realizar predicciones de popularidad para nuevas canciones.

## 1.2 Planteamiento del problema
La gran cantidad de canciones disponibles en plataformas de streaming dificulta identificar qué características musicales se relacionan con su popularidad. Aunque el éxito de una canción también depende de factores externos, este proyecto busca determinar en qué medida las características musicales permiten predecir su nivel de popularidad mediante técnicas de Machine Learning.

## 1.3 Objetivos del proyecto
**Objetivo general**

Desarrollar un modelo de Machine Learning capaz de predecir la popularidad de canciones en Spotify a partir de sus características musicales, así como analizar los factores que influyen en dicha popularidad.

**Objetivos específicos**
* Analizar el conjunto de datos mediante técnicas de análisis exploratorio para comprender su estructura, calidad y relaciones entre variables
* Desarrollar y comparar diferentes modelos de regresión para identificar el modelo con mejor capacidad predictiva de la popularidad de las canciones.
* Indentificar grupos de canciones con características musicales similares mediante técnicas de aprendizaje no supervisado y desarrollar una aplicación interactiva en Shiny para Python para visualizar los principales resultados del proyecto y realizar predicciones.

## 1.4 Criterios de éxito
El éxito del proyecto se evaluará mediente los siguientes criterios:

* Se desarrolla un modelo de regresión evalueado mediante validación cruzada utilizando las métricas MAE, RMSE y R^2
* Se comparan distintos modelos de Machine Learning y se justifica la selección del modelo final con base en su desempeño.
* Se identifican grupos de canciones con características musicales similares que permiten caracterizar diferentes perfiles musicales.
* Se desarrolla una aplicación funcional en Shiny para Python que permita realizar predicciones  y visualizar los resultados del proyecto.

## 1.5 Alcance del proyecto
El presente proyecto comprende el desarrollo y comparación de distintos modelos de regresión para estimar la popularidad de canciones en Spotify a partir de sus característcas musicales. Asimismo, incluye el análisis exploratorio del conjunto de datos, la aplicación de técnicas de clustering para identificar perfiles musicales similares y el desarrollo de una aplicación interactiva en Shiny para Python.

El proyecto no contempla el análisis de letras de canciones, el procesamiento de audio, la construcción de sistemas de recomendación musical ni la predicción de tendencias temporales o del éxito comercial de artistas.

## 1.6 Hipótesis inicial
Se plantea la hipótesis de que es posible estimar, al menos parcialmente, la popularidad de una canción a partir de sus características musicales. Asimismo, se espera que algunos atributos musicales presenten una mayor influencia que otros en la capacidad predictiva del modelo y que existan grupos de canciones con perfiles musicales similares identificables mediente técnicas de clustering.

## 1.7 Metodología
El proyecto se desarrolló siguiendo la metodología CRISP-DM. Cada una de sus etapas se implementó de la siguiente manera:

* Business understanding: Definición del problema, los objetivos y el alcance del proyecto.
* Data Understanding: Exploración del conjunto de datos para comprender sus características, distribución y calidad.
* Data preparation: Prepocesamiento de los datos, incluyendo limpieza, transformación y selección de variables.
* Modeling: Desarollo y comparación de modelos de regresión, así como aplicación de técnicas de clustering para complementar el análisis de los datos.
* Evaluation: Evaluación del desempeño de los modelos mediante métricas apropiadas y selección del modelo final.
* Deployment: Desarrollo de una aplicación en Shiny para Python para visualizar los resultados y realizar predicciones.
