## 2.5 Análisis exploratorio

### 2.5.1 Variable objetivo

```python
plt.hist(df["popularity"], bins=20)
plt.title("Distribution of Popularity")
plt.xlabel("Popularity")
plt.ylabel("Frequency")
plt.show()
```

### 2.5.2 Variables numéricas

```python
numeric_columns = df.select_dtypes(include=["int64", "float64"]).columns

df[numeric_columns].hist(
    figsize=(18, 12),
    bins=20
)

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(4, 4, figsize=(18, 12))

for ax, col in zip(axes.flatten(), numeric_columns):
    ax.boxplot(df[col], vert=True)
    ax.set_title(col)

plt.tight_layout()
plt.show()
```

### 2.5.3 Variables categóricas

```python
df["track_genre"].value_counts()
```

```python
df["explicit"].value_counts()
```

```python
df["artists"].nunique()
```

## 2.6 Relaciones entre variables

```python
plt.figure(figsize=(12,10))

corr = df.select_dtypes(include=["int64","float64"]).corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")
plt.show()
```
