## 2.4 Análisis exploratorio

### 2.4.1 Variable objetivo

```python
plt.hist(df["popularity"], bins=20)
plt.title("Distribution of Popularity")
plt.xlabel("Popularity")
plt.ylabel("Frequency")
plt.show()
```

La distribución de la variable popularity presenta una alta concentración de canciones con valores bajos y una menor frecuencia de canciones muy populares. Esta característica debe considerarse al interpretar los resultados obtenidos por los modelos de regresión.

### 2.4.2 Variables numéricas

```python
numeric_columns = df.select_dtypes(include=["int64", "float64"]).columns

df[numeric_columns].hist(
    figsize=(18, 12),
    bins=20
)

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(4, 4, figsize=(18, 12))

for ax, col in zip(axes.flatten(), numeric_columns):
    ax.boxplot(df[col], vert=True)
    ax.set_title(col)

plt.tight_layout()
plt.show()
```

Las variables key, mode y time_signature se encuentran representadas mediante valores numéricos, sin embargo, corresponden a categorías musicales y, por tanto, no deben interpretarse como variables continuas.

Las variables numéricas presentan diferentes distribuciones. Variables como duration_ms, speechiness, instrumentalness y liveness muestran distribuciones asimétricas y presencia de valores atípicos. En los boxplots se observa la presencia de valores extremos en variables como duration_ms, loudness, liveness, tempo, instrumentalness y speechiness. Dado el dominio del problema, estos valores no se consideran necesariamente errores de captura, sino posibles características válidas de determinadas canciones.

Las variables Unnamed: 0.1 y Unnamed: 0 presentan una distribución uniforme y corresponden únicamente a índices del conjunto de datos, por lo que no aportan información para el análisis.

## 2.5 Relaciones entre variables

```python
plt.figure(figsize=(12,10))

corr = df.drop(columns=["Unnamed: 0.1", "Unnamed: 0"]).select_dtypes(include=["int64","float64"]).corr()

sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Matrix")
plt.show()
```

Las características musicales presentan algunas correlaciones lineales moderadas. Destacan relaciones como la relación positiva entre energy y loudness (0.76), así como las correlaciones negativas de acousticness con energy (-0.73) y loudness (-0.59).

Por otro lado, no se observan correlaciones lineales fuertes entre la variable objetivo popularity y las características analizadas, lo que sugiere que su comportamiento no puede explicarse mediante relaciones lineales simples con una única característica musical.

## 2.6 Hallazgos principales

El análisis exploratorio permitió identificar que la variable objetivo popularity presenta una distribución sesgada hacia valores bajos, mientras que las canciones altamente populares son menos frecuentes.

Las características musicales muestran distribuciones diversas, incluyendo asimetrías y valores atípicos en algunas variables. Asimismo, key, mode y time_signature, aunque están representadas numéricamente, corresponden a variables categóricas.

No se identificaron correlaciones lineales fuertes entre la popularidad y las características musicales analizadas. Sin embargo, sí se observaron relaciones importantes entre algunas variables predictoras, como energy y loudness, y relaciones inversas de acousticness con ambas.

También se identificaron columnas de índice sin valor analítico y una cantidad reducida de valores faltantes, los cuales serán tratados durante la preparación de los datos.
